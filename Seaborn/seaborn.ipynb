```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme()
```

```python
sns.get_dataset_names()
```

```python
tips = sns.load_dataset("tips")
print(tips.head())
```

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker",
    size="size",
    style="smoker"
)
```

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    kind="line"
)
```

```python
# Scatter plot

tips = sns.load_dataset("tips")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="time"
)
```

```python
# line plot

flights = sns.load_dataset("flights")

sns.lineplot(
    data=flights,
    x="year",
    y="passengers"
)
```

```python
# bar plot
sns.barplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex"
)
```
