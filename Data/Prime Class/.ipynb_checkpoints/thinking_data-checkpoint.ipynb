```python
import json
```

```python
# load the data
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)
    return data
```

```python
data = load_data("stroe_data.json")
print(data)
```

```python
# clean & strucure data 
def clean_data(data):
    text_to_num = {"one": 1, "two": 2, "three" :3, "four": 4, "five":5, "six": 6, "seven": 7, "eight": 8}

    for user in data:
        # clean age 
        raw_age = user["age"].strip().lower()
        if(raw_age in text_to_num):
            raw_age = text_to_num[raw_age]
            
        # print(raw_age)
        user["age"] = raw_age
```

```python
clean_data(data)
```
